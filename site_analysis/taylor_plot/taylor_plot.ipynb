```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scienceplots
from matplotlib import cm
from matplotlib.lines import Line2D
```

```python
plt.style.use(['science', 'no-latex', 'retro'])
plt.rc('font', family='Arial')
```

```python
plot_font = 50
mod_marker = ['o', 'o', 'o', 'o']
```

```python
mod_label = ['Slab', 'Urb', 'Urb_Veg', 'CLM5U']

label_sites = ['AU-Preston', 'AU-SurreyHills', 'CA-Sunset', 'FI-Kumpula', 'FI-Torni', 'FR-Capitole',
        'GR-HECKOR', 'JP-Yoyogi', 'KR-Jungnang', 'KR-Ochang', 'MX-Escandon', 'NL-Amsterdam',
        'PL-Lipowa', 'PL-Narutowicza', 'SG-TelokKurau06', 'UK-KingsCollege', 'UK-Swindon', 'US-Baltimore',
        'US-Minneapolis1', 'US-Minneapolis2', 'US-WestPhoenix']
```

```python
def get_var_lists(var_type):
    if var_type == 'turbulent':
        var_mod = ['f_fsena', 'f_lfevpa', 'f_fgrnd']
        var_obs = ['Qh', 'Qle', 'Qg']
        row_titles = ('Sensible Heat', 'Latent Heat', 'Storage Heat')
        outname = 'Fig10'
    elif var_type == 'radiation':
        var_mod = ['f_sr', 'f_olrg', 'f_rnet']
        var_obs = ['SWup', 'LWup', 'Rnet']
        row_titles = ('Upward Shortwave', 'Upward Longwave', 'Net Radiation')
        outname = 'Fig11'
    else:
        raise ValueError("var_type must be 'turbulent' or 'radiation'")
    return var_mod, var_obs, row_titles, outname
```

```python
def get_site_color(site_i):
    if site_i <= 20:
        return cm.tab20(np.linspace(0, 1, 21))[site_i - 1]
    return 'peru'
```

```python
def taylor_diagram(axe):
    axe.set_thetalim(thetamin=0, thetamax=90)
    axe.set_rlim(0, 1.5)

    rad_ticks = [0, 0.25, 0.5, 0.75, 1, 1.25, 1.5]
    axe.set_yticks(rad_ticks)
    axe.set_yticklabels(['0', '0.25', '0.5', '0.75', '1', '1.25', '1.5'], fontsize=plot_font)
    axe.tick_params(axis='y', labelsize=plot_font, pad=20)

    rad_list = [0, 0.2, 0.4, 0.6, 0.7, 0.8, 0.85, 0.9, 0.95, 0.99, 1]
    angle_list = np.rad2deg(np.arccos(rad_list))
    angle_list_rad = np.arccos(rad_list)

    axe.set_thetagrids(angle_list, [], fontsize=plot_font)

    for angle, rad in zip(angle_list, rad_list):
        if rad == 0:
            label_text = '0'
        elif rad == 1:
            label_text = '1'
        else:
            label_text = f'{rad:.2f}'

        axe.text(
            np.deg2rad(angle),
            axe.get_rmax() + 0.11,
            label_text,
            rotation=angle,
            ha='center',
            va='center',
            fontsize=plot_font,
        )

    tick = [axe.get_rmax(), axe.get_rmax() * (1 - 0.02)]
    tick_minor = [axe.get_rmax(), axe.get_rmax() * (1 - 0.01)]

    for angle in angle_list_rad:
        axe.plot([angle, angle], tick, lw=2.5, color='k')
        axe.plot([angle, angle], tick_minor, lw=2.5, color='k')

    for radius, xpos, label in [
        (0.25, 0.095 * np.pi, '0.25'),
        (0.5, 0.175 * np.pi, '0.50'),
        (0.75, 0.269 * np.pi, '0.75'),
        (1.0, 0.36 * np.pi, '1.0'),
    ]:
        circle = plt.Circle(
            (1, 0),
            radius,
            transform=axe.transData._b,
            facecolor=(0, 0, 0, 0),
            edgecolor='grey',
            linestyle='--',
            linewidth=2.5,
        )
        axe.add_artist(circle)
        axe.text(xpos, 1, s=label, fontsize=plot_font, ha='center', va='center', color='grey')

    circle_std = plt.Circle(
        (0, 0),
        1,
        transform=axe.transData._b,
        facecolor=(0, 0, 0, 0),
        edgecolor='k',
        linestyle='-',
        linewidth=1,
    )
    axe.add_artist(circle_std)

    axe.text(0, 1, '\u2605', fontfamily='DejaVu Sans', fontsize=plot_font + 10, ha='center', va='center')
    axe.set_xlabel('Normalized Standard Deviation', labelpad=75, fontsize=plot_font)
    axe.text(np.deg2rad(45), 1.65, s='Correlation', fontsize=plot_font, ha='center', va='bottom', rotation=-45)
```

```python
def draw_taylor(var_type, cache_csv, save=True, outdir='.', dpi=300):
    _, var_obs, row_titles, outprefix = get_var_lists(var_type)

    df = pd.read_csv(cache_csv)
    for col in ['r', 'std_ratio', 'rmse', 'mbe']:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce')

    fig = plt.figure(figsize=(58, 58), dpi=dpi)
    panel_letters = [['a', 'b', 'c', 'd'], ['e', 'f', 'g', 'h'], ['i', 'j', 'k', 'l']]

    for ivar, obs_var in enumerate(var_obs):
        for imod, model in enumerate(mod_label):
            ax = plt.subplot(3, 4, imod + ivar * 4 + 1, projection='polar')
            ax.spines[:].set_linewidth(5)
            taylor_diagram(ax)

            letter = panel_letters[ivar][imod]
            title = r'Urb_{Veg}' if model == 'Urb_Veg' else model
            ax.set_title(rf'$\mathregular{{({letter})~{title}}}$', fontsize=60, y=1.15, weight='bold')

            sub = df[(df['var_type'] == var_type) & (df['var'] == obs_var) & (df['model'] == model)]
            for _, row in sub.iterrows():
                if pd.isna(row['r']) or pd.isna(row['std_ratio']):
                    continue

                theta = np.arccos(np.clip(row['r'], -1, 1))
                site_i = int(row['site_i'])
                color = get_site_color(site_i)

                ax.plot(
                    theta,
                    row['std_ratio'],
                    mod_marker[imod],
                    color=color,
                    markersize=25,
                    label=row['site'],
                    mec='black',
                    mew=2.5,
                )

    legend_handles = [
        Line2D(
            [0],
            [0],
            marker='o',
            linestyle='None',
            color=get_site_color(site_i),
            markeredgecolor='black',
            markeredgewidth=2.5,
            markersize=18,
            label=site,
        )
        for site_i, site in enumerate(label_sites, start=1)
    ]
    fig.legend(
        legend_handles,
        label_sites,
        loc='lower center',
        fontsize=45,
        ncol=7,
        frameon=False,
        bbox_to_anchor=(0.51, 0.025),
    )

    fig.suptitle(row_titles[0], y=0.92, fontsize=75, weight='bold')
    fig.text(0.5, 0.64, row_titles[1], ha='center', va='center', fontsize=75, weight='bold')
    fig.text(0.5, 0.36, row_titles[2], ha='center', va='center', fontsize=75, weight='bold')
    fig.subplots_adjust(hspace=0.3)

    if save:
        outname = f'{outdir}/{outprefix}.pdf'
        plt.savefig(outname, dpi=dpi)
        print('[Saved figure]', outname)

    return fig
```

```python
def main():
    CACHE_FILE = "taylor_metrics.csv"
    draw_taylor('radiation', cache_csv=CACHE_FILE, save=True, outdir='./', dpi=300)
    draw_taylor('turbulent', cache_csv=CACHE_FILE, save=True, outdir='./', dpi=300)
    
    plt.show()
```

```python
if __name__ == '__main__':
    main()
```
